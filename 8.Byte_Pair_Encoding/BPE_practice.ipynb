```python
import importlib
import tiktoken

print("Tiktoken version: ",importlib.metadata.version('tiktoken'))
```

```python
#Available encodings:
print(tiktoken.list_encoding_names())
```

```python
tiktoken=tiktoken.get_encoding('gpt2')
```

```python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
     " of someunknownPlace."
)
integers=tiktoken.encode(text, allowed_special={"<|endoftext|>"})
print("Tokens: ",integers)
```

```python
org_text=tiktoken.decode(integers)
print("Original Text: ",org_text)
```

```python
#Example
integers=tiktoken.encode("Hello,I am Pankaj Kumar.")
print("Tokens: ",integers)
org_text=tiktoken.decode(integers)
print("Original Text: ",org_text)
```
